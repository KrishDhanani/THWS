```python
# 4th example
# Eponentially distributed

import math

x = 1
def get_random_num(a=423, c=1637289, m=2**32):
    global x
    x = (a * x + c) % m
    return x / m

lst = []
for i in range(1000):
    lst.append(get_random_num())
```

```python
# Estimation
est_lst = []

def estimation(i, pre_chosen_lmbd):
    x = - math.log(1-lst[i]) / pre_chosen_lmbd
    est_lst.append(x)

pre_chosen_lmbd = 99
for i in range(1000):
    estimation(i, pre_chosen_lmbd)

# estimated lambda
mean = sum(est_lst) / len(est_lst)
estimated_lmbd = 1 / mean
estimated_lmbd
```

```python
# visulization

import matplotlib.pyplot as plt

plt.hist(est_lst, bins=10, edgecolor='black', color='red')
plt.xlabel("values")
plt.ylabel("frequency")
plt.show()
```
