```python
import csv
import pandas as pd
import numpy as np

# with open("../anthrokids/anthrokids.csv", 'r') as file:
#     data = csv.DictReader(file)
#     print(data)

    # for row in data:
    #     print(row['height'])

df = pd.read_csv('../anthrokids/anthrokids.csv')
data_height = df['height']

data_height = data_height.round()
# print(data_height) # it still showing the float data type we need to convert it to int
```

```python
data_height = data_height.dropna()
data_height = data_height.astype(int)
data_height  # Type of numpy int32
```

```python
x = pd.Series(data_height)
y = x.value_counts(normalize=True).sort_values()   # normalize automatically counts pro. for us
print(type(y))
y
```

```python
import random

def get_height():

    cum_sum = 0
    r_variable = random.uniform(0,1) # evry nnum. has equal chance

    for height, prob in y.items():
        cum_sum += prob
        if cum_sum >= r_variable:   # checking is our com_pro is greter than r_variable
            print(f"Sample height: {height}")
            return height
            # break
        
```

```python
# Getting Samples

sample_list = []
for i in range(999):
    sample_height = get_height()
    sample_list.append(sample_height)

# print(sample_list)
```

```python
generated_data = pd.Series(sample_list)
ge_data = generated_data.value_counts(normalize=True).sort_values()
ge_data
```

```python
import matplotlib.pyplot as plt

plt.hist(y, bins=25, density=True, histtype='bar', edgecolor='black', color='red')
plt.xlabel("probability")
plt.ylabel("height of student")
# plt.x
plt.show()
```
