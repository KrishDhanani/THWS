```python
# 2nd exercise

# random number gerneration

x = 1
def get_next_number(a=423, c=1637289, m=2**32):
    global x
    next_x = (a * x + c) % m
    x = next_x
    return next_x / m

lst_num = []
for i in range(999):
    lst_num.append(get_next_number())

# lst_num
```

```python
min(lst_num)
```

```python
max(lst_num)
```

```python
# My observation says if i have only 5 values in my sample then it is worse 
# bcs it is more far then 0 &1 on the other hand when i have more samples in my list i can see that it is more better 
# because its max and min are close to 0 and 1
```

```python
import matplotlib.pyplot as plt

plt.hist(lst_num, bins=5, edgecolor='black', color='red')
plt.xlabel("random numbers")
plt.ylabel("frequency")
plt.show()
```

```python
# If the bars are all roughly the same height (flat on top like a brick), then your generator is working perfectly.
# If there are huge "towers" or "gaps," your parameters ($a, c, m$) might be bad.
```
