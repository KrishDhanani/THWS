```python
# Exercise 3
# 
# The most common way to do this from scratch is the Box-Muller Transform. 
# It takes two independent 'uniform random numbers' (u1 and u2) between 0 and 1 and transforms them into two independent, 
# normally distributed numbers z0 & z1.

import numpy as np
import math

x = 1
def get_next_num(a=1664525, c=1013904223, m=2**32):
    global x
    next_x = (a * x + c) % m
    x = next_x
    
    return next_x / m    # for 0 to 1 scale 
```

```python
# applying formula of box mular

def get_normal_sample():
    u1 = get_next_num()
    u2 = get_next_num()
    # print(u1,u2)

    # Box mullar logic
    z0 = (math.sqrt(-2 * math.log(u1))) * math.cos(2 * math.pi * u2)
    return z0

# creating sample list
lst = []
for i in range(1000):
    lst.append(get_normal_sample())

lst
```

```python
# mean
n = len(lst)
mean = sum(lst) / n 

# standard deviation
n_lst = []
for i in lst:
    n_lst.append((i-mean)**2)
    
N = len(n_lst)
variance = sum(n_lst)/N
s_d = math.sqrt(variance)

print(f"The {mean=} and {s_d=}")
```

```python
# Even though we used a formula designed to produce a mean of 0 and a std dev of 1, 
# the sample estimates are not exactly 0 and 1. There is a slight offset.

# The accuracy is high, but not perfect. You can calculate the Error as:
# {Mean Error} = 0 - {your\_mean}
# {Std Dev Error} = 1.0 - \text{your_sd}

# If you change range(1000) to range(10000), your estimates will get much closer to the "True" parameters (0 and 1). 
# This is because a larger sample size reduces sampling bias.
```
