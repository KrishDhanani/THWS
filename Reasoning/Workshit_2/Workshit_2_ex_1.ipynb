```python
import numpy as np 
```

```python

def get_data(N, dimensions):
    return np.random.normal(loc=0.0, scale=1.0, size=[N,dimensions])

data = get_data(100000, 2)
data
```

```python
estimated_mean = np.mean(data, axis=0)
estimated_mean
```

```python
estimated_cov = np.cov(m=data, rowvar=False)
estimated_cov
```

```python
# So after changing the N values i observed that as many data you have your estimated_mean and 
# estimated_cov is getting more accurate like mean close to 0 and cov is 1. 

# Sigma =  Var(X) & Cov(X, Y) 
#           Cov(Y, X) & Var(Y)
# this is what is we got as output using np.cov
```
