```python
from sklearn.datasets import load_iris
import numpy as np

iris = load_iris() 
iris_feat = iris.data
X = np.ones((iris_feat.shape[0], 4))
X[:, 1:4] = iris_feat[:, :3]
y = iris_feat[:,3]
X, y
XTX = X.T @ X
XTy = X.T @ y
beta_hat =  np.linalg.solve(XTX, XTy)
# print(f'{beta_hat=}')
y_hat = X @ beta_hat
# print(f'{y_hat=}') 

y_pred = beta_hat[0] + beta_hat[1] * 5 + beta_hat[2] * 3.5 + beta_hat[3] * 1.5
y_pred
```
